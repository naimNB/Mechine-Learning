### Import Liberary

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(0)

df = pd.DataFrame(
    data={
        'Feature' : np.arange(1,101),
        'Target' : np.array([0]*60 + [1]*40)
    }
)
```

```python
df.head()
```

```python
sns.lmplot(
    data=df,
    x='Feature',
    y='Target'
)
```

```python
df_jitter = df.copy()

df_jitter['Target'] = df_jitter["Target"] + np.random.normal(0, 0.30, size=len(df))
df_jitter["Target"] = df_jitter["Target"].clip(0, 1)
```

```python
plt.figure(figsize=(8,5))
sns.regplot(
    x="Feature",
    y="Target",
    data=df_jitter,
    label="Liner Regression"
)

sns.regplot(
    x="Feature",
    y="Target",
    data=df_jitter,
    logistic=True,
    color="red",
    
)

plt.legend()
plt.show()

```
