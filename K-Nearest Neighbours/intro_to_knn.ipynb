```python
import numpy as np
import pandas as pd

np.random.seed(4)


#classs A
A_ax1 = np.random.normal(loc=3, scale=0.3, size=10)
A_ax2 = np.random.normal(loc=3, scale=0.3, size=10)

#classs B
B_ax1 = np.random.normal(loc=4, scale=0.3, size=10)
B_ax2 = np.random.normal(loc=4, scale=0.3, size=10)



df = pd.DataFrame(
    {
        "Feature_1":np.concatenate([A_ax1, B_ax1]),
        "Feature_2":np.concatenate([A_ax2, B_ax2]),
        "Target" : ["Class_A"] * 10 + ["Class_B"] * 10
    }
)

df.head()
```
