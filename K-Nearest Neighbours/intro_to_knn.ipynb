```python
import numpy as np
import pandas as pd

np.random.seed(4)


#classs A
A_ax1 = np.random.normal(loc=3, scale=0.3, size=10)
A_ax2 = np.random.normal(loc=3, scale=0.3, size=10)

#classs B
B_ax1 = np.random.normal(loc=4, scale=0.3, size=10)
B_ax2 = np.random.normal(loc=4, scale=0.3, size=10)



df = pd.DataFrame(
    {
        "Feature_1":np.concatenate([A_ax1, B_ax1]),
        "Feature_2":np.concatenate([A_ax2, B_ax2]),
        "Target" : ["Class_A"] * 10 + ["Class_B"] * 10
    }
)

df.head()
```

```python
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
sns.scatterplot(
    data=df,
    x="Feature_1",
    y="Feature_2",
    s=100,#size of the figure
    hue="Target"

)
```

```python
plt.figure(figsize=(6,5))

sns.scatterplot(
    data=df,
    x="Feature_1",
    y="Feature_2",
    hue="Target",
    s=80
)

sns.scatterplot(
    x=[3.5],
    y=[3.5],
    c="black",
    s=200,
    markers="x",
    label='new point'
)

plt.legend()
```

```python
import numpy as np
```

```python
data_point = np.array([3.5,3.5])

df['distance'] = np.sqrt(
    (df['Feature_1']-data_point[0])**2 + 
    (df['Feature_2']-data_point[0])**2
)


df
```

```python
df.sort_values(
    by="distance",
    inplace=True
)

df
```
